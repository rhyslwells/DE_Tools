```python
import pandas as pd

# Creating a sample DataFrame
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

# ### Selecting Columns

# You can select a single column using bracket notation or dot notation
column_a = df['A']  # Using bracket notation
print("Column A using bracket notation:\n", column_a)

column_b = df.A  # Using dot notation
print("\nColumn B using dot notation:\n", column_b)

# ### Selecting Rows by Index

# You can select rows by their index position using slicing
rows_0_to_2 = df[0:3]  # Selects the first three rows (index 0, 1, 2)
print("\nRows 0 to 2:\n", rows_0_to_2)

# ### Selecting Rows by Date Range

# If your DataFrame has a DateTime index, you can select rows within a specific date range
date_rng = pd.date_range(start='2013-01-01', end='2013-01-06', freq='D')
df_dates = pd.DataFrame(date_rng, columns=['date'])
df_dates.set_index('date', inplace=True)

# Selecting rows by date range
selected_dates = df_dates['2013-01-02':'2013-01-04']
print("\nSelected dates (2013-01-02 to 2013-01-04):\n", selected_dates)

# ### Label-based Selection

# Use `.loc` or `.at` to select rows by label
df_weather = pd.DataFrame({'Weather': ['Sunny', 'Rain', 'Cloudy'], 'Temp': [30, 22, 25]})
df_weather.set_index('Weather', inplace=True)

# Selecting a specific row by label
rain_row = df_weather.loc['Rain']  # or df_weather.at['Rain']
print("\nRow for 'Rain' label:\n", rain_row)

# ### Position-based Selection

# Use `.iloc` or `.iat` to select rows by position
third_row = df_weather.iloc[2]  # Selects the third row
print("\nThird row by position:\n", third_row)

specific_value = df_weather.iat[1, 1]  # Selects the value at row 1, column 1
print("\nSpecific value (row 1, column 1):\n", specific_value)

# ### Conditional Selection

# Select rows based on a condition
filtered_df = df[df['A'] > 0]  # Selects rows where column 'A' is greater than 0
print("\nFiltered DataFrame where 'A' > 0:\n", filtered_df)

# Create a new DataFrame based on a condition
df_new = df[df['A'] >= 2]  # Selects rows where 'A' is greater than or equal to 2
print("\nFiltered DataFrame where 'A' >= 2:\n", df_new)
```

```python
# Create a sample DataFrame with a date range as index
dates = pd.date_range("20130101", periods=6)
df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list("ABCD"))

print("Original DataFrame:")
print(df)
```

```python
# Selection of Data
print("\nSelecting column A (Two Methods):")
print(df['A'])  # Selecting column 'A' using dictionary-like access
print("-----")
print(df.A)     # Selecting column 'A' using attribute access
```

```python
print("\nSelecting rows by slicing (first three rows):")
print(df[0:3])  # Select first three rows using slicing
```

```python
# Because we are using the date as an index
print("\nSelecting rows using date range:")
print(df['20130102':'20130104'])  # Select rows by date range
```

```python
# Selecting using loc and at (label-based selection)
df.loc['Rain'] = [1, 2, 3, 4]  # Adding a row with index 'Rain'
print("\nSelecting row 'Rain' using loc:")
print(df.loc['Rain'])  # Select the row with label 'Rain'
```

```python
print("\nSelecting a single value using at:")
print(df.at['Rain', 'B'])  # Select value at row 'Rain' and column 'B'
```

```python
# Selecting using iloc and iat (position-based selection)
print("\nSelecting the fourth row using iloc:")
print(df.iloc[3])  # Select the fourth row (index-based)
```

```python
# Sorting Data

print("\nSorting by column 'C' in descending order:")
print(df.sort_values(by='C', ascending=False))
```

```python
# updating values

df['Hobbyist'].map({'Yes': True, 'No': False})

# updating a specific value
df.loc[2, 'last'] = 'Smith'

# filtering then updating all to the same
filt = (df['email'] == 'JohnDoe@email.com')
df[filt]['last'] = 'Smith'
df.loc[filt, 'last'] = 'Smith'

# mapping values with a dict:
df['first'] = df['first'].replace({'Corey': 'Chris', 'Jane': 'Mary'})

# apply function to column
def update_email(email):
    return email.upper()
df['email'].apply(update_email)
```
