```python
# Importing Libraries
import pandas as pd
import numpy as np
```

```python
# Creating a Sample DataFrame for Basic Selection
basic_df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print("Sample DataFrame:")
print(basic_df)
```

# Column and Row Selection

```python
## Selecting Columns Using Bracket and Dot Notation
column_a = basic_df['A']  # Bracket notation
column_b = basic_df.A  # Dot notation
print("Column A using bracket notation:")
print(column_a)
print("\nColumn B using dot notation:")
print(column_b)
```

```python
## Selecting Rows by Index Range
rows_0_to_2 = basic_df[0:3]  # First three rows
print("\nRows 0 to 2:")
print(rows_0_to_2)
```

```python
# Creating a Sample DataFrame for Date-based Selection
date_rng = pd.date_range(start='2013-01-01', end='2013-01-06', freq='D')
date_df = pd.DataFrame(index=date_rng, columns=['A', 'B', 'C'])
print(date_df)
```

```python
# Selecting Rows by Date Range
selected_dates = date_df['2013-01-02':'2013-01-04']
print("\nSelected dates (2013-01-02 to 2013-01-04):")
print(selected_dates)
```

# Label-based and Position-based Selection

```python
## Creating a DataFrame for Label-based and Position-based Selection
df_weather = pd.DataFrame({'Weather': ['Sunny', 'Rain', 'Cloudy'], 'Temp': [30, 22, 25]})
df_weather.set_index('Weather', inplace=True)
print(df_weather)
```

```python

## Using `.loc` for Label-based Selection
rain_row = df_weather.loc['Rain']
print("\nRow for 'Rain' label:")
print(rain_row)
```

```python

## Using `.iloc` for Position-based Selection
third_row = df_weather.iloc[2]
print("\nThird row by position:")
print(third_row)
```

```python

specific_value = df_weather.iat[1, 0]  # Select value at row 1, column 0
print("\nSpecific value (row 1, column 0):")
print(specific_value)
```

# Conditional Selection

```python
filtered_df = basic_df[basic_df['A'] > 0]
print("\nFiltered DataFrame where 'A' > 0:")
print(filtered_df)
```

```python
# Creating a Random DataFrame for Numeric Operations
dates = pd.date_range("20130101", periods=6)
numeric_df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list("ABCD"))
print("Original DataFrame:")
print(numeric_df)
```

```python
# Boolean Indexing (Filtering Data)
print("\nFiltering rows where column 'A' > 0:")
print(numeric_df[numeric_df.A > 0.2])
```

```python
## Advanced Filtering
filt = (numeric_df['A'] > 0.2) & (numeric_df['B'] > 1)
print(numeric_df[filt])
```

```python
# Sorting Data
print("\nSorting by column 'C' in descending order:")
print(numeric_df.sort_values(by='C', ascending=False))
```

# String Operations

```python
## Creating a DataFrame for String Operations
string_df = pd.DataFrame({'full_name': ['John Doe', 'Jane Smith', 'Chris Brown'],
                          'email': ['john@example.com', 'jane@example.com', 'chris@example.com'],
                          'Hobbyist': ['Yes', 'No', 'Yes']})
print(string_df)
```

```python
## Splitting a Column
string_df[['first', 'last']] = string_df['full_name'].str.split(' ', expand=True)
print("\nSplitting full_name into first and last:")
print(string_df[['first', 'last']])
```


# Updating Values


```python
string_df['Hobbyist'] = string_df['Hobbyist'].map({'Yes': True, 'No': False})
print("\nUpdated DataFrame:")
print(string_df)
```

```python
string_df.loc[2, 'last'] = 'Smith'  # Updating a specific value
print("\nUpdated DataFrame:")
print(string_df)
```

```python

filt = (string_df['first'] == 'John')
string_df.loc[filt, 'last'] = 'Smith'
print("\nUpdated DataFrame:")
print(string_df)
```

```python
string_df['first'] = string_df['first'].replace({'Corey': 'Chris', 'Jane': 'Mary'})
print("\nUpdated DataFrame:")
print(string_df)
```

```python
def update_email(email):
    return email.upper()

string_df['email'] = string_df['email'].apply(update_email)

print("\nUpdated DataFrame:")
print(string_df)
```
