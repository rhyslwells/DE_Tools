```python
import pandas as pd
from IPython.display import display
```

# Concatenate

```python
# Combine two DataFrames row-wise (default axis=0)
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
print("Before:")
print(df1, "\n")
print(df2)
print("Combining two DataFrames row-wise:")
concat_df = pd.concat([df1, df2], ignore_index=True)
print(concat_df, "\n")
print("Concatenate along columns (axis=1)")
result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
df3 = pd.DataFrame({'A': [9, 10], 'C': [11, 12]})

# Concatenate DataFrames with different columns
print(df1)
print(df3)
result = pd.concat([df1, df3], axis=0)
print(result)
```

# Merge

```python
# Merge on a shared column 'key'
df1 = pd.DataFrame({'key': ['A', 'B'], 'value': [1, 2]})
df2 = pd.DataFrame({'key': ['A', 'B'], 'value': [3, 4]})
print("Merging on shared column 'key':")
print("Before:")
print(df1)
print(df2)
merged_df = pd.merge(df1, df2, on='key')
print("After:")
print(merged_df, "\n")
```

```python
# Merge on shared key
df_12 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6], 'key': [1, 2, 3]})
df_13 = pd.DataFrame({'A': [7, 8, 9,10], 'B': [10, 11, 12,11], 'key': [1, 2, 3,4]})
print("Merging on shared 'key' column:"
    "\nBefore:")
print(df_12,"\n")
print(df_13, "\n")
print(pd.merge(df_12, df_13, on='key', how='inner'), "\n")
```

```python
# Merge on 'city'
df1 = pd.DataFrame({"city": ["new york", "chicago", "orlando"], "temperature": [21, 14, 35]})
df2 = pd.DataFrame({"city": ["chicago", "new york", "orlando"], "humidity": [65, 68, 75]})
print("Merging on column 'city':")
print("Before:")
print(df1)
print(df2)
merged = pd.merge(df1, df2, on="city")
print("After:")
print(merged, "\n")
```

```python


# Merge with indicator
print("Merge with indicator column:")
df1 = pd.DataFrame({"city": ["new york", "chicago", "orlando"], "temperature": [21, 14, 35]})
df2 = pd.DataFrame({"city": ["chicago", "new york", "orlando"], "humidity": [65, 68, 75]})
print("Merging on column 'city':")
print("Before:")
print(df1)
print(df2)
print(pd.merge(df1, df2, on="city", how="outer", indicator=True), "\n")

# Merge with suffixes
df1 = pd.DataFrame({
    "city": ["new york", "chicago", "orlando", "baltimore"],
    "temperature": [21, 14, 35, 38],
    "humidity": [65, 68, 71, 75]
})
df2 = pd.DataFrame({
    "city": ["chicago", "new york", "san diego"],
    "temperature": [21, 14, 35],
    "humidity": [65, 68, 71]
})
print("Merge with custom suffixes for overlapping columns:")
print(pd.merge(df1, df2, on="city", how="outer", suffixes=('_first', '_second')), "\n")

```

### SQL-style joins

```python
df1 = pd.DataFrame({"city": ["new york", "chicago", "orlando", "baltimore"], "temperature": [21, 14, 35, 38]})
df2 = pd.DataFrame({"city": ["chicago", "new york", "san diego"], "humidity": [65, 68, 71]})
print(df1)
print(df2)
```

```python
print("Inner join on 'city':")
print(pd.merge(df1, df2, on="city", how="inner"), "\n")
```

```python
print("Outer join on 'city':")
print(pd.merge(df1, df2, on="city", how="outer"), "\n")
```

```python
print("Left join on 'city':")
print(pd.merge(df1, df2, on="city", how="left"), "\n")
```

```python
print("Right join on 'city':")
print(pd.merge(df1, df2, on="city", how="right"), "\n")
```

# Join

```python
df1 = pd.DataFrame({'val1': [1, 2]}, index=['A', 'B'])
df2 = pd.DataFrame({'val2': [3, 4]}, index=['A', 'B'])
print(df1)
print(df2)
# Join on index
df1.join(df2)
```

```python
### 3. `.join()` with `on=` joins index with a column of another DataFrame
df1 = pd.DataFrame({'val1': [1, 2]}, index=['A', 'B'])
df2 = pd.DataFrame({'key': ['A', 'B'], 'val2': [3, 4]})
print(df1)
print(df2)
df1.join(df2.set_index('key'), on=df1.index)
```

```python
# Join two DataFrames using their index
df1 = pd.DataFrame({'key': ['A', 'B'], 'val1': [1, 2]}).set_index('key')
df2 = pd.DataFrame({'key': ['A', 'B'], 'val2': [3, 4]}).set_index('key')
print("Joining two DataFrames on their index:")
print("Before:")
print(df1)
print(df2)
joined_df = df1.join(df2, lsuffix='_left', rsuffix='_right')
print("After:")
print(joined_df, "\n")
```

```python
# Join on index
df1 = pd.DataFrame({"city": ["new york", "chicago", "orlando"], "temperature": [21, 14, 35]}).set_index('city')
df2 = pd.DataFrame({"city": ["chicago", "new york", "orlando"], "humidity": [65, 68, 75]}).set_index('city')
print(df1)
print(df2)
print("\n Joining on index after setting 'city' as index:")
print(df1.join(df2, lsuffix='_l', rsuffix='_r'), "\n")
```

```python
# Join with different indexes
df_12 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]}, index=[1, 2, 3])
df_13 = pd.DataFrame({'C': [7, 8, 9], 'D': [10, 11, 12]}, index=[1, 4, 5])
print(df_12)
print(df_13)
print("Joining DataFrames with partially overlapping indices:")
print(df_12.join(df_13, how='outer'), "\n")
```
