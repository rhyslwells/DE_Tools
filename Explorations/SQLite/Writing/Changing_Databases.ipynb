Here we show how to:

- Remove Table
- Rename Table
- Add Column
- Rename Column

```python
import sqlite3
import shutil
import pandas as pd

# Define database paths
original_db_path = "../Datasets/database/longlist.db"
copied_db_path = "Outputs/long_list_changed.db"

# Make a copy of the original database
shutil.copy(original_db_path, copied_db_path)
print("Database copied to 'Outputs/long_list_changed.db'.")

# Connect to the copied database
conn = sqlite3.connect(copied_db_path)
cursor = conn.cursor()
```

```python
# 1. Show existing tables before modifications
existing_tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("Existing tables before modification:")
display(existing_tables)
```

```python
# 2. Remove the 'ratings' table if it exists
cursor.execute("DROP TABLE IF EXISTS ratings;")
print("Table 'ratings' removed (if it existed).")

tables_after_drop = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("Tables after removing 'ratings':")
display(tables_after_drop)
```

```python
# 3. Rename the 'publishers' table to 'book_publishers'
cursor.execute("ALTER TABLE publishers RENAME TO book_publishers;")
print("Table 'publishers' renamed to 'book_publishers'.")

tables_after_rename = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("Tables after renaming 'publishers' to 'book_publishers':")
display(tables_after_rename)
```

```python
# 4. Add a 'genre' column to the 'books' table
cursor.execute("ALTER TABLE books ADD COLUMN genre TEXT;")
print("Column 'genre' added to 'books'.")

books_columns = pd.read_sql_query("PRAGMA table_info(books);", conn)
print("Columns of 'books' table after adding 'genre':")
display(books_columns)
```

```python
# 5. Rename the 'format' column to 'book_format' in 'books' (SQLite 3.25+ required)
cursor.execute("ALTER TABLE books RENAME COLUMN format TO book_format;")
print("Column 'format' renamed to 'book_format' in 'books'.")

books_columns_after_rename = pd.read_sql_query("PRAGMA table_info(books);", conn)
print("Columns of 'books' table after renaming 'format' to 'book_format':")
display(books_columns_after_rename)
```

```python
# Commit changes and close connection
conn.commit()
conn.close()
print("Database operations completed.")
```
